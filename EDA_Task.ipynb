```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
sns.set(font_scale=1.0)

pd.set_option('display.max_columns',None)
```

```python
df = pd.read_csv("application_data.csv")

print(df.shape)

df.head()
```

```python
missing = pd.DataFrame({
    "Missing Values":df.isnull().sum(),
    "Percentage":round(df.isnull().mean()*100,2)
})

missing = missing.sort_values("Percentage",ascending=False)

missing.head(30)
```

```python
df.isnull().sum()
```

```python
missing = pd.DataFrame({
    "Missing Values":df.isnull().sum(),
    "Percentage":round(df.isnull().mean()*100,2)
})

missing = missing.sort_values("Percentage",ascending=False)

missing.head(30)
```

```python
threshold = 50

drop_cols = missing[missing["Percentage"]>threshold].index

df.drop(columns=drop_cols,inplace=True)

print(df.shape)
```

```python
df["LOAN_INCOME_RATIO"] = df["AMT_CREDIT"] / df["AMT_INCOME_TOTAL"]
```

```python
df["EMI_INCOME_RATIO"] = df["AMT_ANNUITY"] / df["AMT_INCOME_TOTAL"]
```

```python
df["CREDIT_GOODS_RATIO"] = df["AMT_CREDIT"] / df["AMT_GOODS_PRICE"]
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df["LOAN_INCOME_RATIO"],bins=50)

plt.xlim(0,10)

plt.show()
```

```python
education = df.groupby("NAME_EDUCATION_TYPE")["TARGET"].mean()*100

education = education.sort_values()

education.plot(kind="bar",figsize=(10,5))

plt.ylabel("Default %")

plt.show()
```

```python
occupation = df.groupby("OCCUPATION_TYPE")["TARGET"].mean()*100

occupation.sort_values().plot(kind="barh",figsize=(10,8))

plt.xlabel("Default %")

plt.show()
```

```python
corr = df.select_dtypes(include=np.number).corr()

plt.figure(figsize=(18,12))

sns.heatmap(corr,
            cmap="coolwarm",
            center=0)

plt.show()
```
